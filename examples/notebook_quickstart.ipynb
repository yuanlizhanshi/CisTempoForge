# CisTempoForge notebook quickstart
Load a typed config, change it interactively, then train into a caller-owned directory.

```python
from cistempoforge import load_config, fit, load_model
config = load_config('d0_d5_config.json')
config.training.learning_rate = 1e-4
config.training.epochs = 40
```

```python
result = fit(config, output_dir='../runs/d0_d5_seed42')
result.history.tail()
```

```python
model, checkpoint = load_model(result.best_checkpoint, device='cuda')
checkpoint['metrics']
```
